# shapefile of ERW

```python
import xarray as xr
import numpy as np
import rasterio.features
import geopandas as gpd
from shapely.geometry import shape

# 1. Open dataset and load the mask variable (assumes 'mask' is 2D)
ds = xr.open_dataset(ERW_TOPO_FILE)
mask = ds['mask'].values  # numpy array

# 2. Get coordinates info
# Assumes 2D mask with coordinates 'y' and 'x' or 'lat' and 'lon'
# Example: get 1D arrays of coordinates
x = ds['x'].values  # or ds['lon'].values
y = ds['y'].values  # or ds['lat'].values

# 3. Build an affine transform from the coordinates
# Assumes x and y are regularly spaced
from affine import Affine

# Calculate resolution and origin
x_res = (x[-1] - x[0]) / (len(x) - 1)
y_res = (y[-1] - y[0]) / (len(y) - 1)
# Affine transform: (top-left x, pixel width, 0, top-left y, 0, pixel height)
transform = Affine.translation(x[0] - x_res / 2, y[0] - y_res / 2) * Affine.scale(x_res, y_res)

# 4. Extract polygons from the mask
# We extract polygons for regions where mask == 1 (True)
mask_bool = mask.astype(bool)

# rasterio.features.shapes returns (geom, value)
shapes_generator = rasterio.features.shapes(mask_bool.astype(np.uint8), transform=transform)

# Filter shapes where value==1
polygons = [shape(geom) for geom, value in shapes_generator if value == 1]

# 5. Create GeoDataFrame with CRS
# Get CRS from dataset attributes or define manually if known (e.g. EPSG:4326)
crs = ds.attrs.get('crs', 'EPSG:26913')  # Replace if no CRS attribute or set your own

gdf = gpd.GeoDataFrame({'geometry': polygons}, crs=crs)

# Optional: save to a shapefile or GeoJSON
# gdf.to_file('mask_polygons.shp')  # or .to_file('mask_polygons.geojson', driver='GeoJSON')

print(f"Extracted {len(gdf)} polygons from mask.")
```
